```python
# vectorization

arr=np.array([1,2,3,4,5],dtype="int64")
arr2=np.array([6,7,8,9,10])
print(arr+arr2)
print(arr**arr2)
print(arr+1)
```

```python
#Usage of Numpy Broadcasting -- Std Normalization(z-score)

arr=np.array([[1,2],[3,4]])
mean=np.mean(arr)
print(f"mean = {mean}")
std_deviation=np.std(arr)
print(f"standard deviation = {std_deviation}")
normalized_arr=(arr-mean)/std_deviation
print(normalized_arr)
```
