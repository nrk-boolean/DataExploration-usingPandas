```python
#loading the data
def load_data(filename):
    with open(filename,"r") as f:
        data=json.load(f)
    return data 
```

```python
data=load_data("store_data.json")
print(data,"\n\n\n")
print(type(data))
```

```python
#clean & structure the Data
def clean_data(data):
    text_to_number={"one":1,"two":2,"three":3,"four":4,"five":5}
    cleaned_data=[] #for final cleaned Data
    unique_users=set()
    
    for user in data:
        # #cleaning rating
        raw_rating=str(user["rating"]).strip().lower()
        if(raw_rating in text_to_number):
            raw_rating=text_to_number[raw_rating]
        user["rating"]=raw_rating
        
        #Handling Missing Values
        raw_age=user.get("age")
        if(raw_age==None):
            user["age"]=None

        #if user is present,then continue
        if(user["name"].strip() in unique_users):
            continue
        
        #if not add
        cleaned_data.append(user)
        unique_users.add(user["name"])
    
    return cleaned_data
```

```python
clean_data(data) #Before replacing
```

```python
print(clean_data(data))
cleaned_data=clean_data(data)
```

```python
# Building recommendation Features
def get_recommendations(data):
    recommendations=[]

    for user in data:
        current_recommendation={}
        current_recommendation["name"]=user["name"]
        if(float(user["rating"])>=4):
            current_recommendation["Brand"]="Apple"
        else:
            current_recommendation["Brand"]="Samsung"
        recommendations.append(current_recommendation)
    return recommendations
```

```python
recommendations=get_recommendations(cleaned_data)
print(recommendations)
```
